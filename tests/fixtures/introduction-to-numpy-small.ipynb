```python
import numpy as np
print(f"NumPy Version: {np.__version__}")
```

## 1. Creating Arrays

The primary data structure in NumPy is the `ndarray`. Let's create a 1D array (vector) and inspect its properties.

```python
a1 = np.array([10, 20, 30])
print("Array:", a1)
print("Shape:", a1.shape)
print("Type:", type(a1))
```

## 2. Array Operations

One of the key features of NumPy is vectorization, allowing operations on entire arrays without explicit loops.

```python
a2 = a1 * 2
print("Multiplied by 2:", a2)
```
